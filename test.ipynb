## 7. Insert product information into products collection
This cell inserts sample product data into the `products` collection.

Each product has an `item_id` and a corresponding `product_name`, which will later be used for joining with the sales data.

```python
db.sales.distinct("cat_id")
```

```python
import random

db.products.delete_many({})

category_names = {
    "FOODS": {
        "names": [
            "Bread", "Milk", "Rice", "Eggs", "Banana",
            "Apple", "Cheese", "Pasta", "Yogurt", "Orange Juice"
        ],
        "weights": [25, 20, 18, 15, 10, 8, 7, 6, 5, 4]
    },

    "HOUSEHOLD": {
        "names": [
            "Detergent", "Toilet Paper", "Dish Soap", "Tissues",
            "Trash Bags", "Shampoo", "Laundry Powder",
            "Bleach", "Sponges", "Air Freshener"
        ],
        "weights": [20, 18, 15, 14, 10, 8, 6, 5, 3, 2]
    },

    "HOBBIES": {
        "names": [
            "Board Game", "Toy Car", "Puzzle", "Art Kit",
            "Coloring Book", "Football", "Doll",
            "Chess Set", "Lego Blocks", "Painting Set"
        ],
        "weights": [20, 18, 15, 14, 10, 8, 6, 4, 3, 2]
    }
}


items = db.sales.find({}, {"item_id": 1, "cat_id": 1})

products_docs = []
seen = set()

for doc in items:
    item_id = doc["item_id"]
    cat = doc["cat_id"]

    if item_id in seen:
        continue
    seen.add(item_id)

    if cat in category_names:
        names = category_names[cat]["names"]
        weights = category_names[cat]["weights"]

        # weighted random
        product_name = random.choices(
            names,
            weights=weights,
            k=1
        )[0]
    else:
        product_name = "Unknown Product"

    products_docs.append({
        "item_id": item_id,
        "product_name": product_name
    })

db.products.insert_many(products_docs)

print("Done! Added", len(products_docs), "products.")
```

## 8. Join sales with products (Lookup pipeline)
This cell uses MongoDB aggregation with `$lookup` to join the `sales` collection with the `products` collection based on `item_id`.

Then it extracts the product name and removes unnecessary nested fields.

This allows us to map each product ID to its real product name.

```python
pipeline = [
    {
        "$lookup": {
            "from": "products",
            "localField": "item_id",
            "foreignField": "item_id",
            "as": "product_info"
        }
    },
    {
        "$unwind": "$product_info"
    },
    {
        "$addFields": {
            "product_name": "$product_info.product_name"
        }
    },
    {
        "$project": {
            "product_info": 0
        }
    }
]

result = list(db.sales.aggregate(pipeline))

for doc in result[:5]:
    print(doc)
```

## 9. Create Virtual Warehouse per Store
This cell builds a virtual warehouse for each store using aggregation.

First, it joins sales with product names, then calculates total sales per product per store.

Finally, it groups all products under each store to simulate a warehouse structure.

```python
pipeline = [
    # 1) join products
    {
        "$lookup": {
            "from": "products",
            "localField": "item_id",
            "foreignField": "item_id",
            "as": "product_info"
        }
    },

    {
        "$unwind": "$product_info"
    },

    # 2) add product name
    {
        "$addFields": {
            "product_name": "$product_info.product_name"
        }
    },

    # 3) calculate quantity (بدل total_sales)
    {
        "$group": {
            "_id": {
                "store_id": "$store_id",
                "item_id": "$item_id",
                "product_name": "$product_name"
            },
            "quantity": {
                "$sum": {
                    "$add": [
                        "$d_1713",
                        "$d_1714",
                        "$d_1715",
                        "$d_1716",
                        "$d_1717"
                    ]
                }
            }
        }
    },

    # 4) build warehouse per store
    {
        "$group": {
            "_id": "$_id.store_id",
            "warehouse": {
                "$push": {
                    "item_id": "$_id.item_id",
                    "product_name": "$_id.product_name",
                    "quantity": "$quantity"
                }
            }
        }
    },

    # 5) final output
    {
        "$project": {
            "_id": 0,
            "store_id": "$_id",
            "warehouse": 1
        }
    }
]

result = list(db.sales.aggregate(pipeline))

# for doc in result[:1]:
#     print(doc)
```

```python
for doc in result:
    print("="*40)
    print("STORE:", doc["store_id"])
    print("Total items:", len(doc["warehouse"]))

    total_qty = sum(item["quantity"] for item in doc["warehouse"])
    print("Total quantity in store:", total_qty)

    top_items = sorted(doc["warehouse"], key=lambda x: x["quantity"], reverse=True)[:5]

    print("\nTop 5 products:")
    for item in top_items:
        print(item["product_name"], "->", item["quantity"])
```
